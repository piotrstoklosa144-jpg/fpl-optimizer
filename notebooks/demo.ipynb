## 2. Using the Python API

Top 15 players by expected points over the next 5 gameweeks.

```python
from fpl_optimizer import add_expected_points, load_fpl_data

data = load_fpl_data(horizon=5)
players = add_expected_points(data)

cols = ["web_name", "position", "team_name", "cost", "next_opponent", "xp_next", "xp_total"]
players.sort_values("xp_total", ascending=False)[cols].head(15).round(2)
```

Best value per million, per position:

```python
players["xp_per_m"] = players["xp_total"] / players["cost"]
(players[players["x_mins"] > 60]
    .sort_values("xp_per_m", ascending=False)
    .groupby("position").head(3)[cols + ["xp_per_m"]]
    .round(2))
```
